```python
def forward_PSF(microscope, grid, aberration):
    _, _, psf = microscope.compute_PSF(grid, aberration)
    psf = cp.asarray(psf, dtype=cp.float64) ** (1/3)
    return psf / cp.sum(psf)


def circular_convolve(f, psf):
    return cp.real(cp.fft.ifft2(cp.fft.fft2(f) * cp.fft.fft2(cp.fft.ifftshift(psf))))


def forward_image(microscope, grid, f, aberration):
    psf = forward_PSF(microscope, grid, aberration)
    return circular_convolve(f, psf)


def add_gaussian_noise(image, SNR, rng):
    std_dev_noise = cp.std(image) / cp.sqrt(cp.asarray(SNR, dtype=image.dtype))
    return image + rng.normal(0, std_dev_noise, image.shape, dtype=image.dtype)
```

```python
SNR = 1000
rng_np = np.random.default_rng(15)
rng_cp = cp.random.default_rng(15)

true_aberration = generate_johnson_aberration(2.0, alpha, rng = rng_np)
print(true_aberration.strengths)

bias_modes = [[-2, 2], [0, 2], [2, 2], [0, 4], [-4, 4], [4, 4]]
bias_strength = 0.1
a_stack = [EmptyAberration()] + [Aberration([m], [bias_strength]) for m in bias_modes]

d_stack = cp.stack([
    forward_image(microscope, grid, f, true_aberration + a)
    for a in a_stack
], axis=0)
#d_stack = add_gaussian_noise(d_stack, SNR, rng_cp)
```

```python
import time
```

```python
def objective(microscope, grid, d_stack, a_proposed, a_stack, gamma):
    S_stack = cp.stack([
        cp.fft.fft2(forward_PSF(microscope, grid, a_proposed + a))
        for a in a_stack
    ], axis=0)
    D_stack = cp.stack([
        cp.fft.fft2(d)
        for d in d_stack
    ], axis=0)

    
    first_term = cp.sum(cp.abs(D_stack) ** 2)
    numerator = cp.abs(cp.sum(cp.conj(S_stack) * D_stack, axis=0)) ** 2
    denominator = gamma + cp.sum(cp.abs(S_stack) ** 2, axis=0)

    return float(cp.asnumpy(cp.real(first_term - cp.sum(numerator / denominator))))
```

```python
def objective_batched(
    microscope,
    grid,
    D_stack,
    first_term,
    a_proposed,
    a_stack,
    gamma,
):
    aberrations = [a_proposed + a for a in a_stack]

    _, _, I_stack_jax = microscope.compute_PSF_stack_jax(
        grid,
        aberrations,
    )

    I_stack = cp.asarray(I_stack_jax)
    S_stack = cp.fft.fft2(I_stack, axes=(-2, -1))

    inner = cp.sum(cp.conj(S_stack) * D_stack, axis=0)
    power = cp.sum(cp.abs(S_stack) ** 2, axis=0)

    J = first_term - cp.sum(cp.abs(inner) ** 2 / (gamma + power))

    return float(cp.asnumpy(cp.real(J)))

def optimize(microscope, grid, d_stack, a_stack, modes_corrected, gamma, c_guess = 0):
    if c_guess == 0:
        print("here")
        c_guess = np.zeros(len(modes_corrected))
    d_stack_gpu = cp.asarray(d_stack)
    D_stack = cp.fft.fft2(d_stack_gpu, axes=(-2, -1))
    first_term = cp.sum(cp.abs(D_stack) ** 2)

    iter_counter = {"n": 0}
    last_J = {"value": None}

    def single_arg_objective(c_guess):
        J = objective_batched(
            microscope,
            grid,
            D_stack,
            first_term,
            Aberration(modes_corrected, np.asarray(c_guess, dtype=np.float64)),
            a_stack,
            gamma,
        )
        last_J["value"] = J
        return J

    for c in [0.0, 0.25, 0.5, 0.75, 1.0]: 
        t = time.time() 
        single_arg_objective(np.array([c] * len(modes_corrected))) 
        s = time.time() 
        print(s-t)

    def callback(c_guess):
        iter_counter["n"] += 1
        print(f"iter {iter_counter['n']:4d}: J = {last_J['value']:.6e}")
        print(f"    c = {c_guess}")

    return minimize(
        single_arg_objective,
        x0=c_guess,
        method="Nelder-Mead",
        bounds=Bounds(-.1, .1),
        options={
            "disp": True,
            "maxiter": 500,
            "xtol": 1e-4,
            "ftol": 1e-10,
        },
        callback=callback,
    )
```

```python
modes_corrected = true_aberration.modes
result = optimize(microscope, grid, d_stack, a_stack, modes_corrected, 1e-6)
result
```
